```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn import linear_model

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error
```

1.Objective is to train three linear regression models and study the relation between the median house value  and three independent variables (Income,population and number of household) out of which the goal is to know which variable best predicts the median house value.

```python
# Data loading 
data = pd.read_csv('housing.csv')
data.head()
```

```python
# Data described
data.describe()
```

```python
sns.scatterplot(x=data['median_income'], y=data['median_house_value'])
plt.title('Median House Value vs. Median Income')
plt.show()
```

```python
sns.scatterplot(x=data['population'], y=data['median_house_value'])
plt.title('Median House Value vs. Population')
plt.show()
```

```python
sns.scatterplot(x=data['households'], y=data['median_house_value'])
plt.title('Median House Value vs. Number of Households')

plt.show()
```

Variables Used above are as Follows:

1.Median House Value :- The median house value ranges from 14999 to 500001

2.Median Income:- It showcases the median income between the range of 0-15

3.Population:- This variable showcases the population around each blocks. We also can see some outliers in the variable 

4.Household :- This variable shows the households in each pricerange. The highest of which lies in around 6000

```python
train_size = int(len(data) * 0.8)
train = data[:train_size]
test = data[train_size:]
```

```python
model = linear_model.LinearRegression()
```

```python
model.fit(X=train[["median_income"]], y=train[["median_house_value"]])
```

```python
print(f"Intercept: {model.intercept_}")
print(f"Coefficients: {model.coef_}")
```

```python
predictions = model.predict(test[["median_income"]])

mse = mean_squared_error(test["median_house_value"], predictions)
mae = mean_absolute_error(test["median_house_value"], predictions)

print(f"Mean Squared Error (MSE): {mse}")
print(f"Mean Absolute Error (MAE): {mae}")
```

```python
sns.scatterplot(x="median_income", y="median_house_value", data=train, color="blue")
plt.plot(train[["median_income"]], model.predict(train[["median_income"]]), color="black", linewidth=3)
plt.show()
```
